```python
#creation 1D Numpy Array

import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)

#Inspection and Properties:

print("Shape:", temperatures_w1.shape)

print("Data Type:", temperatures_w1.dtype)

print("Number of Elements:", temperatures_w1.size)

#Array Operation
#a Convert Celsius into Fahrenheit
fahrenheit = (temperatures_w1 * 9/5) + 32

print("Temperature in Fahrenheit:", fahrenheit)

#b Find maximum temperature

print("Maximum Temperature:", np.max(temperatures_w1))

#minimum temperature

print("Minimum Temperature:", np.min(temperatures_w1))

#mean temperature
print("Mean Temperature:", np.mean(temperatures_w1))
#Array Slicing and Indexing:

#a.Extract first three days
print(temperatures_w1[:3])
#b.Extract weekend temperatures
print(temperatures_w1[-2:])
#Extract middle three days
print(temperatures_w1[2:5])



#Create a 2D Array:
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
#Inspect and Slice the 2D Array
print("Shape:", temperatures.shape)
#Data type
print("Data Type:", temperatures.dtype)
#Total number of elements
print("Total Elements:", temperatures.size)
#Extract Week 1 temperatures
print(temperatures[0])
#Extract Week 2 temperatures
print(temperatures[1])
#Extract weekend temperatures for both weeks
print(temperatures[:, -2:])

#Import Pandas
import pandas as pd
#Create a Pandas Series
marks = pd.Series(  [95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
#Indexing and Slicing
# Access 1st rank using integer position
print(marks.iloc[0])

# Retrieve top 3 ranks using loc
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Retrieve 3rd rank using iloc
print(marks.iloc[2])

# Boolean masking: marks greater than 90
print(marks[marks > 90])
#Manipulating Pandas Series
marks.loc['Rank1'] = 100
print(marks)
#Remove the last rank
marks = marks.drop('Rank5')
print(marks)
#Calculate CGPA
cgpa = marks / 10
print(cgpa)

#Create a DataFrame

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
#Data Exploration
print(transactions)
#Display the first 5 rows using head()
print(transactions.head())
#Display the last 5 rows using tail()
print(transactions.tail())
#Find the shape
print(transactions.shape)
#Display column names
print(transactions.columns)
#Check data types
print(transactions.dtypes)
#Display basic information
transactions.info()


#Select Columns
#Display only ProductCategory and Amount
print(transactions[['ProductCategory', 'Amount']])
#Retrieve the last 3 columns using iloc
print(transactions.iloc[:, -3:])

#Filter Rows
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)]
print(filtered_data)
#Value Counts
print(transactions['ProductCategory'].value_counts())
#Find Unique Values
print(transactions['Region'].unique())

#Group By and Calculate Mean

print(transactions.groupby('Region')['Amount'].mean())


#Manipulating Pandas DataFrame
#Modify Amount for TransactionID 102
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)

#Add a Discount Column
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)
#Remove TransactionID 109
transactions = transactions.drop(
    transactions[transactions['TransactionID'] == 109].index)
print(transactions)
#Delete the Discount Column
transactions = transactions.drop('Discount', axis=1)
print(transactions)
```
